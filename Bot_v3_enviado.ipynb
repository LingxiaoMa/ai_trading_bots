```python
def compute_ema(prices, window):
    alpha = 2 / (window + 1)
    ema = [prices[0]]

    for i in range(1, len(prices)):
        value = alpha * prices[i] + (1 - alpha) * ema[i-1]
        ema.append(value)

    return np.array(ema)
```

```python
def compute_sma(prices, window):
    sma = []

    for i in range(len(prices)):
        start = max(0, i - window + 1)
        value = np.mean(prices[start:i+1])
        sma.append(value)

    return np.array(sma)
```

```python
def compute_lma(prices, window):
    lma = []
    weights = np.arange(1, window + 1)

    for i in range(len(prices)):
        if i < window:
            current_prices = prices[:i+1]
            current_weights = np.arange(1, len(current_prices) + 1)
        else:
            current_prices = prices[i-window+1:i+1]
            current_weights = weights

        value = np.sum(current_prices * current_weights) / np.sum(current_weights)
        lma.append(value)

    return np.array(lma)
```

```python
def compute_ma(prices, window, ma_type):
    if ma_type == "EMA":
        return compute_ema(prices, window)
    elif ma_type == "SMA":
        return compute_sma(prices, window)
    elif ma_type == "LMA":
        return compute_lma(prices, window)
    else:
        raise ValueError("Invalid moving average type")
```

A flexible fitness function is used to support multiple trading strategies.
The type of moving average (EMA, SMA, LMA) is passed as a parameter. This allows the same pipeline to evaluate different strategies, including mixed approaches such as EMA short vs SMA long.

```python
def fitness(particle, prices, short_type, long_type):
    short_window = int(round(particle[0]))
    long_window = int(round(particle[1]))

    if short_window < 2 or long_window < 2:
        return 0
    if short_window >= long_window:
        return 0
    if long_window >= len(prices):
        return 0

    ma_short = compute_ma(prices, short_window, short_type)
    ma_long = compute_ma(prices, long_window, long_type)

    signals = generate_signals(ma_short, ma_long)
    final_cash, _ = backtest(prices, signals)

    return final_cash
```

```python
def generate_signals(ma_short, ma_long):
    signals = ["HOLD"]

    for i in range(1, len(ma_short)):
        if ma_short[i] > ma_long[i] and ma_short[i-1] <= ma_long[i-1]:
            signals.append("BUY")
        elif ma_short[i] < ma_long[i] and ma_short[i-1] >= ma_long[i-1]:
            signals.append("SELL")
        else:
            signals.append("HOLD")

    return signals
```

```python
def backtest(prices, signals, initial_cash=1000, fee=0.03):
    cash = initial_cash
    btc = 0.0
    portfolio_values = []

    for i in range(len(signals)):
        price = prices[i]
        signal = signals[i]

        if signal == "BUY" and cash > 0:
            btc = (cash * (1 - fee)) / price
            cash = 0

        elif signal == "SELL" and btc > 0:
            cash = btc * price * (1 - fee)
            btc = 0

        portfolio_value = cash if btc == 0 else btc * price
        portfolio_values.append(portfolio_value)

    if btc > 0:
        cash = btc * prices[-1] * (1 - fee)
        btc = 0
        portfolio_values[-1] = cash

    return cash, portfolio_values
```

```python
def pso_optimize(prices, short_type, long_type,
                 n_particles=8, n_iterations=15,
                 short_min=2, short_max=50,
                 long_min=10, long_max=200,
                 w=0.7, c1=1.5, c2=1.5):

    positions = np.zeros((n_particles, 2))
    velocities = np.zeros((n_particles, 2))

    for i in range(n_particles):
        short_w = np.random.uniform(short_min, short_max)
        long_w = np.random.uniform(long_min, long_max)

        while short_w >= long_w:
            short_w = np.random.uniform(short_min, short_max)
            long_w = np.random.uniform(long_min, long_max)

        positions[i] = [short_w, long_w]

    pbest_positions = positions.copy()
    pbest_scores = np.array([
        fitness(p, prices, short_type, long_type) for p in positions
    ])

    gbest_index = np.argmax(pbest_scores)
    gbest_position = pbest_positions[gbest_index].copy()
    gbest_score = pbest_scores[gbest_index]

    history = []

    for iteration in range(n_iterations):
        for i in range(n_particles):
            r1, r2 = np.random.rand(), np.random.rand()

            velocities[i] = (
                w * velocities[i]
                + c1 * r1 * (pbest_positions[i] - positions[i])
                + c2 * r2 * (gbest_position - positions[i])
            )

            positions[i] += velocities[i]

            positions[i][0] = np.clip(positions[i][0], short_min, short_max)
            positions[i][1] = np.clip(positions[i][1], long_min, long_max)

            if positions[i][0] >= positions[i][1]:
                positions[i][0] = max(short_min, positions[i][1] - 1)

            score = fitness(positions[i], prices, short_type, long_type)

            if score > pbest_scores[i]:
                pbest_scores[i] = score
                pbest_positions[i] = positions[i].copy()

            if score > gbest_score:
                gbest_score = score
                gbest_position = positions[i].copy()

        history.append(gbest_score)
        print(f"Iter {iteration+1}: Best = {gbest_score:.2f}")

    return gbest_position, gbest_score, history
```
